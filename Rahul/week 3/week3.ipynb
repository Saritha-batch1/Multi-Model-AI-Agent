```python
import pandas as pd
df = pd.read_csv("contextual_analysis.csv")
df["Synthesized_Finding"] = (
    "Condition: " + df["Medical Condition"]
    + " | Pattern: " + df["Condition_Pattern"]
    + " | Risk: " + df["Contextual_Risk"]
)
df.to_csv("synthesis.csv", index=False)
print("Synthesis completed")
df.head()
```

```python
df = pd.read_csv("risk_assessment.csv")
def generate_recommendation(condition):
    if condition == "Diabetes":
        return "Monitor blood glucose, maintain balanced diet"
    elif condition == "Hypertension":
        return "Reduce salt intake, manage stress"
    elif condition == "Cancer":
        return "Follow oncologist guidance and treatment plan"
    elif condition == "Asthma":
        return "Avoid allergens, use inhaler as prescribed"
    else:
        return "Maintain healthy lifestyle"
df["Recommendation"] = df["Medical Condition"].apply(generate_recommendation)
df.to_csv("recommendation_generation.csv", index=False)
print("Recommendations generated")
df.head()
```
