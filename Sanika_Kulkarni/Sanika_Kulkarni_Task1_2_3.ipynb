```python
import re
from nltk.corpus import stopwords
from nltk import word_tokenize
from collections import Counter
from nltk.util import ngrams

STOP = set(stopwords.words('english'))

def clean_text_basic(text):
    # normalize whitespace, remove weird control chars
    text = text.replace('\r',' ').replace('\t',' ')
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()
    return text

def tokenize_and_filter(text):
    text_l = text.lower()
    # remove punctuation except % and - and . inside numbers
    text_l = re.sub(r'[^a-z0-9%.\- ]+', ' ', text_l)
    tokens = word_tokenize(text_l)
    tokens = [t for t in tokens if t.isalnum() or (any(c.isdigit() for c in t) and '%' in t)==False]
    tokens = [t for t in tokens if t not in STOP and len(t)>1]
    return tokens
```

```python
cbc_clean = clean_text_basic(cbc_text)
full_clean = clean_text_basic(full_text)
ocr_clean = clean_text_basic(ocr_text)   # 3rd PDF
print("Cleaning Done")
```

```python
import re

parameter_pattern = re.compile(
    r"(?P<param>(?:[A-Za-z][A-Za-z \/\(\)\-]*))\s+"
    r"(?P<value>\d+\.?\d*)\s*"
    r"(?P<flag>Low|High|Borderline|L\\|H\\|L|H|Critical)?\s*"
    r"(?P<range>\d+\.?\d*\s*[-–]\s*\d+\.?\d*)?\s*"
    r"(?P<unit>[a-zA-Z/%µcuL]+)?",
    flags=re.IGNORECASE
)
```

```python
def extract_parameters(text):
    results = []

    for m in parameter_pattern.finditer(text):
        param = m.group("param").strip()

        # Reject NON-medical lines
        if len(param.split()) > 4:
            continue
        if any(x in param.lower() for x in ["lab", "pathology", "medical", "address", "page", "report"]):
            continue

        results.append({
            "param": param,
            "value": m.group("value"),
            "flag": m.group("flag"),
            "range": m.group("range"),
            "unit": m.group("unit")
        })

    return results
```

```python
cbc_params = extract_parameters(cbc_clean)
full_params = extract_parameters(full_clean)
ocr_params = extract_parameters(ocr_clean)

print("CBC Parameters Found:", len(cbc_params))
print("Full Body Report Parameters Found:", len(full_params))
print("OCR Report Parameters Found:", len(ocr_params))
```

```python
import pandas as pd

def build_df(params, source):
    rows = []
    for p in params:
        rows.append({
            "Parameter": p["param"],
            "Value": p["value"],
            "Flag": p["flag"],
            "Reference Range": p["range"],
            "Unit": p["unit"],
            "Source": source
        })
    return rows

all_rows = []
all_rows += build_df(cbc_params, "CBC_Report")
all_rows += build_df(full_params, "Full_Body_Report")
all_rows += build_df(ocr_params, "OCR_CBC_Report")

df = pd.DataFrame(all_rows)
df
```

```python
# List of real medical keywords
medical_keywords = [
    "hemoglobin", "hb", "rbc", "wbc", "platelet", "neutrophils",
    "lymphocytes", "monocytes", "basophils", "eosinophils",
    "mcv", "mch", "mchc", "rdw", "pcv", "hematocrit",
    "glucose", "cholesterol", "hdl", "ldl", "triglycerides"
]

# Function to check if row contains medical parameter
def is_medical(row):
    p = str(row["Parameter"]).lower()
    return any(key in p for key in medical_keywords)

# Apply filtering
clean_df = df[df.apply(is_medical, axis=1)]

# Remove empty values
clean_df = clean_df[clean_df["Value"].notnull()]

clean_df
```

```python
import re
import pandas as pd

cleaned_rows = []

def clean_unit(unit):
    if not unit: return None
    unit = unit.lower().strip()

    replacements = {
        "g/dl": "g/dL",
        "mg/dl": "mg/dL",
        "fl": "fL",
        "pg": "pg",
        "cumm": "cells/uL",
        "%": "%",
        "10^3/uL": "cells/uL"
    }

    for old, new in replacements.items():
        if old in unit:
            return new
    return unit


def clean_parameter_name(name):
    if not name: return None
    name = str(name)

    # Remove extra symbols
    name = re.sub(r"[^A-Za-z0-9 %/().-]", " ", name)

    # remove repeated words
    words = name.split()
    seen = set()
    filtered = []
    for w in words:
        if w.lower() not in seen:
            filtered.append(w)
            seen.add(w.lower())

    name = " ".join(filtered)

    # Map long names to standard short
    mapping = {
        "hemoglobin": "Hemoglobin",
        "blood hemoglobin": "Hemoglobin",
        "rbc count": "RBC Count",
        "wbc count": "WBC Count",
        "platelet count": "Platelet Count",
        "mean corpuscular volume": "MCV",
        "mean corpuscular hemoglobin": "MCH",
        "mean corpuscular hemoglobin concentration": "MCHC",
        "red cell distribution width": "RDW"
    }

    name_lower = name.lower()
    for key in mapping:
        if key in name_lower:
            return mapping[key]

    return name.strip()


def clean_value(value):
    if value is None: return None

    # Remove commas
    value = str(value).replace(",", "")

    # Extract number
    match = re.findall(r"\d+\.?\d*", value)
    if match:
        try:
            return float(match[0])
        except:
            return None
    return None


# ---- APPLY CLEANING ----
for _, row in clean_df.iterrows():
    cleaned_rows.append({
        "Parameter": clean_parameter_name(row["Parameter"]),
        "Value": clean_value(row["Value"]),
        "Flag": row["Flag"],
        "Reference Range": row["Reference Range"],
        "Unit": clean_unit(row["Unit"]),
        "Source": row["Source"]
    })

validated_df = pd.DataFrame(cleaned_rows)

validated_df
```
