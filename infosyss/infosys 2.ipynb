```python
import pandas as pd
data = {"age":[45,60,35,55],"total_cholesterol":[190,240,180,260],"hdl":[50,40,55,35],"ldl":[120,160,110,180],"systolic_bp":[120,145,118,150]}
df = pd.DataFrame(data)
df["lipid_ratio"] = df["total_cholesterol"] / df["hdl"]
df["bp_risk"] = df["systolic_bp"].apply(lambda x: 1 if x >= 140 else 0)
df["cholesterol_risk"] = df["lipid_ratio"].apply(lambda x: 1 if x >= 5 else 0)
df["age_risk"] = df["age"].apply(lambda x: 1 if x >= 50 else 0)
df["cardio_risk_score"] = df["bp_risk"] + df["cholesterol_risk"] + df["age_risk"]
df["risk_level"] = df["cardio_risk_score"].map({0:"Low",1:"Low",2:"Moderate",3:"High"})
df.to_csv("model2_risk_assessment_output.csv", index=False)
df
```

```python
import pandas as pd
data = {"age":[45,60,35,55],"gender":["male","female","female","male"],"lipid_ratio":[3.8,6.0,3.2,6.5],"systolic_bp":[120,145,118,150]}
df = pd.DataFrame(data)
df["age_factor"] = df["age"].apply(lambda x: 1 if x >= 50 else 0)
df["gender_factor"] = df["gender"].apply(lambda x: 1 if x == "male" else 0)
df["lipid_risk"] = df.apply(lambda r: 1 if (r["lipid_ratio"] >= 5 and r["gender"] == "male") or (r["lipid_ratio"] >= 5.5 and r["gender"] == "female") else 0, axis=1)
df["bp_risk"] = df.apply(lambda r: 1 if (r["systolic_bp"] >= 140 and r["age"] >= 50) else 0, axis=1)
df["contextual_risk_score"] = df["age_factor"] + df["gender_factor"] + df["lipid_risk"] + df["bp_risk"]
df["risk_level"] = df["contextual_risk_score"].apply(lambda x: "Low" if x <= 1 else "Moderate" if x == 2 else "High")
df.to_csv("model3_contextual_risk_output.csv", index=False)
df
```

```python
import pandas as pd
data = {"age":[45,60,35,55],"gender":["male","female","female","male"],"total_cholesterol":[190,240,180,260],"hdl":[50,40,55,35],"systolic_bp":[120,145,118,150]}
df = pd.DataFrame(data)
df["model1_cholesterol_flag"] = df["total_cholesterol"].apply(lambda x: 1 if x > 200 else 0)
df["model1_bp_flag"] = df["systolic_bp"].apply(lambda x: 1 if x >= 140 else 0)
df["lipid_ratio"] = df["total_cholesterol"] / df["hdl"]
df["model2_risk_score"] = df["lipid_ratio"].apply(lambda x: 1 if x >= 5 else 0) + df["model1_bp_flag"]
df["model3_context_score"] = df.apply(lambda r: 1 if (r["age"] >= 50 or r["gender"] == "male") else 0, axis=1)
df["final_risk_score"] = df["model1_cholesterol_flag"] + df["model2_risk_score"] + df["model3_context_score"]
df["final_risk_level"] = df["final_risk_score"].apply(lambda x: "Low" if x <= 1 else "Moderate" if x == 2 else "High")
df.to_csv("integrated_model_output.csv", index=False)
df
```
