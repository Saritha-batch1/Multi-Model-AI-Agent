```python
import pandas as pd
import json
import fitz
import os
def extract(file):
    ext = os.path.splitext(file)[1]
    if ext == ".csv":
        df = pd.read_csv(file)
        df.to_csv("output.csv", index=False)
        return df
    if ext == ".json":
        df = pd.read_json(file)
        df.to_csv("output.csv", index=False)
        return df
    if ext == ".pdf":
        text = ""
        for page in fitz.open(file):
            text += page.get_text()
        df = pd.DataFrame({"text": text.split("\n")})
        df.to_csv("output.csv", index=False)
        return df
data = extract("healthcare_dataset.csv")
data.head()
```

```python
import re
import pandas as pd
from collections import Counter
text = """
Healthcare data analysis is important.
Healthcare systems generate large amounts of data.
"""
text = text.lower()
text = re.sub(r'[^\w\s]', '', text)
words = text.split()
word_count = len(words)
unique_words = len(set(words))
common_words = Counter(words).most_common(5)
print("Total words:", word_count)
print("Unique words:", unique_words)
print("Most common words:", common_words)
df = pd.DataFrame(common_words, columns=["Word", "Frequency"])
df.to_csv("text_analysis_output.csv", index=False)
df.head()
```

```python
import re
import pandas as pd
text = """
Healthcare data analysis is important.
Healthcare systems generate large amounts of data.
"""
text = text.lower()
text = re.sub(r'[^\w\s]', '', text)
tokens = text.split()
df = pd.DataFrame(tokens, columns=["tokens"])
df.to_csv("text_processing_output.csv", index=False)
df.head()
```

```python
import re
import pandas as pd
text = """
Patient Age: 45
Blood Pressure: 120/80
Heart Rate: 72
Cholesterol: 190
"""
pairs = re.findall(r'([\w\s]+):\s*([\w\/]+)', text)
df = pd.DataFrame(pairs, columns=["Parameter", "Value"])
df.to_csv("key_parameters_output.csv", index=False)
df
```

```python
import pandas as pd
data = {"Parameter":["Patient Age","Blood Pressure","Heart Rate","Cholesterol"],"Value":["45","120/80","72","190"]}
df = pd.DataFrame(data)
df["Value"] = df["Value"].astype(str)
df["Valid"] = df["Value"].apply(lambda x: x.strip() != "" and x.lower() != "nan")
df.to_csv("data_validation_output.csv", index=False)
df
```

```python
import pandas as pd
data = {"Parameter":["patient age","Blood pressure "," heart rate","Cholesterol level"],"Value":[" 45 ","120 / 80","72 bpm"," 190 mg/dl"]}
df = pd.DataFrame(data)
df["Parameter"] = df["Parameter"].str.lower().str.strip()
df["Value"] = df["Value"].str.lower().str.strip()
df["Value"] = df["Value"].str.replace("bpm","",regex=False)
df["Value"] = df["Value"].str.replace("mg/dl","",regex=False)
df["Value"] = df["Value"].str.replace(" ","",regex=False)
df.to_csv("standardized_data_output.csv", index=False)
df
```
